```python
import torch
```

```python
x = torch.tensor(3., requires_grad=True) #make it true when you want gradient of it
y = x**2
y.backward() #does differentiation
print(x.grad)
```

```python
y #grad_fn=<PowBackward0> just stores in memory that this is an power function and we need to do backprpogation like that
```
