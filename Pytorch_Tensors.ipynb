mathematical operations

```python
#scalar operation
x = torch.rand(2,3)
```

```python
print(x+2)
print(x-2)
print(x*2)
print(x/2)
print(x**2)
```

```python
#element wise operations
a = torch.rand(2,3)
b = torch.rand(2,3)
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**b)
```

```python
f = torch.randint(size = (2,3), low = 0, high = 10, dtype = torch.int32)
g = torch.randint(size = (3,2), low = 0, high = 10, dtype = torch.int32)
print(f)
print(g)
```

```python
c = torch.tensor([1,2,3])
d = torch.tensor([1,2,3])
torch.matmul(c,d)
```

inplace operations

```python
m = torch.rand(2,3)
n = torch.rand(2,3)
```

```python
m + n # doing this will create new tensor which in return occupy some memory instead below we will do the inplace operations
```

```python
n.add_(m)
```

```python
m.add_(n)
```

```python
#adding and _ after the function name will perform inplace operations
```

```python
torch.relu_(m)
```
